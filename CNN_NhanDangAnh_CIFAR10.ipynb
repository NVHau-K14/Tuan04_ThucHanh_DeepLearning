```python
import pandas as pd
import numpy as np
import tensorflow as tf
import keras
# Thêm BatchNormalization, RandomFlip, RandomRotation vào import
from keras.layers import Conv2D, Dense, MaxPool2D, Flatten, Dropout, BatchNormalization, RandomFlip, RandomRotation
from keras.models import Sequential
from keras import Input
import matplotlib.pyplot as plt
import cv2
from google.colab import files

# --- 1. TẢI VÀ CHUẨN BỊ DỮ LIỆU ---
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()

class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
               'dog', 'frog', 'horse', 'ship', 'truck']

# Chuẩn hóa dữ liệu về khoảng [0, 1]
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Chuyển nhãn sang dạng One-hot encoding
num_class = 10
y_train_cat = keras.utils.to_categorical(y_train, num_class)
y_test_cat = keras.utils.to_categorical(y_test, num_class)

# --- 2. XÂY DỰNG MÔ HÌNH CNN ---
input_shape = (32, 32, 3)

model = Sequential()
model.add(Input(shape=input_shape))

# Tăng cường dữ liệu (Data Augmentation) ngay trong mô hình
model.add(RandomFlip("horizontal"))
model.add(RandomRotation(0.1))

# Block 1
model.add(Conv2D(32, kernel_size=(3, 3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(Conv2D(32, kernel_size=(3, 3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(MaxPool2D(pool_size=(2, 2)))
model.add(Dropout(0.25)) # Tắt ngẫu nhiên 25% nơ-ron

# Block 2
model.add(Conv2D(64, kernel_size=(3, 3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(Conv2D(64, kernel_size=(3, 3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(MaxPool2D(pool_size=(2, 2)))
model.add(Dropout(0.35))

# Block 3
model.add(Conv2D(128, kernel_size=(3, 3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(MaxPool2D(pool_size=(2, 2)))
model.add(Dropout(0.45))

# Block phân loại (Dense)
model.add(Flatten())
model.add(Dense(units=128, activation='relu'))
model.add(BatchNormalization())
model.add(Dropout(0.5))
model.add(Dense(units=10, activation='softmax'))

model.summary()

# --- 3. HUẤN LUYỆN MÔ HÌNH ---
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Do có Dropout và Augmentation nên mô hình sẽ học chậm hơn một chút,
# ta tăng epochs lên 30 để mô hình có đủ thời gian "ngấm" kiến thức
model_fit = model.fit(X_train, y_train_cat, batch_size=64, epochs=20,
                      validation_split=0.1, verbose=1)

# --- 4. ĐÁNH GIÁ MÔ HÌNH ---
score = model.evaluate(X_test, y_test_cat, verbose=0)
print(f'\nTest loss: {score[0]:.4f}')
print(f'Test accuracy: {score[1]:.4f}')

# Vẽ đồ thị loss và accuracy
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(model_fit.history['accuracy'], label='Train Acc')
plt.plot(model_fit.history['val_accuracy'], label='Val Acc')
plt.title('Model Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(model_fit.history['loss'], label='Train Loss')
plt.plot(model_fit.history['val_loss'], label='Val Loss')
plt.title('Model Loss')
plt.legend()
plt.show()

```

```python
# --- 5. UPLOAD ẢNH VÀ DỰ ĐOÁN ---
print("\n--- TEST VỚI ẢNH CỦA BẠN ---")
uploaded = files.upload()

for filename in uploaded.keys():
    # Đọc ảnh màu (BGR) và chuyển sang RGB
    img_raw = cv2.imread(filename)
    img_rgb = cv2.cvtColor(img_raw, cv2.COLOR_BGR2RGB)

    # Cắt và Resize về 32x32
    h, w, _ = img_rgb.shape
    min_dim = min(h, w)
    start_x = (w // 2) - (min_dim // 2)
    start_y = (h // 2) - (min_dim // 2)
    img_cropped = img_rgb[start_y:start_y+min_dim, start_x:start_x+min_dim]

    img_32x32 = cv2.resize(img_cropped, (32, 32), interpolation=cv2.INTER_AREA)

    # Chuẩn hóa và reshape
    img_normalized = img_32x32.astype("float32") / 255.0
    img_ready = img_normalized.reshape(1, 32, 32, 3)

    # Dự đoán
    prediction = model.predict(img_ready)
    predicted_class = np.argmax(prediction)
    confidence = np.max(prediction) * 100

    # Hiển thị
    plt.imshow(img_32x32)
    plt.title(f"Dự đoán: {class_names[predicted_class]} ({confidence:.2f}%)")
    plt.axis('off')
    plt.show()
```
